```python
import numpy as np
import matplotlib.pyplot as plt

def objective_function(x):
    """
    The function we want to minimize: f(x) = x^2
    """
    return x**2

def gradient(x):
    """
    The derivative of the function: f'(x) = 2x
    """
    return 2 * x

def gradient_descent(start_x, learning_rate, iterations):
    """
    Performs gradient descent to find the minimum of the function.
    """
    x = start_x
    history = []  # To store the path for visualization

    for i in range(iterations):
        history.append(x)
        # The core update rule: x = x - learning_rate * derivative
        grad = gradient(x)
        x = x - learning_rate * grad
        
        print(f"Iteration {i+1}: x = {x:.4f}, f(x) = {objective_function(x):.4f}")
    
    return x, history

# --- Hyperparameters ---
initial_x = 10.0       # Starting point
eta = 0.1             # Learning rate (Step size)
num_iterations = 20   # Number of steps

# --- Execution ---
final_x, x_history = gradient_descent(initial_x, eta, num_iterations)

print(f"\nFound minimum at x = {final_x:.4f}")

# --- Visualization ---
# This part shows the professional side of your work
x_vals = np.linspace(-11, 11, 100)
y_vals = objective_function(x_vals)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'k-', label='f(x) = x^2') # The function
plt.plot(x_history, [objective_function(val) for val in x_history], 'ro-', label='Optimization Path') # The path
plt.title("Gradient Descent Optimization Process")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```
